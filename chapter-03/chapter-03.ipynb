## 3.1 The Parts of a Chart

### Getting set up

````python
import matplotlib.pyplot as plt
import pandas as pd

plt.style.use('book.mplstyle')
````

````python
sales = pd.read_csv('monthly_sales.csv')
sales.head()
````

## 3.2 Two Ways to Draw a Chart

### `plt.plot()`

````python
plt.plot(sales['month'], sales['sales_2025'])
plt.title('Monthly sales, 2025')
plt.show()
````

### `plt.show()`

Shown in the book but not run here:

```python
plt.plot(sales['month'], sales['sales_2025'])
```

### `plt.subplots()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025')
plt.show()
````

````python
fig, ax = plt.subplots(figsize=(5.5, 2.2))
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025')
plt.show()
````

### `plt.figure()`

````python
plt.figure(figsize=(5.5, 2.2))
plt.plot(sales['month'], sales['sales_2024'])
plt.title('Monthly sales, 2024')
plt.show()
````

## 3.3 Titles, Labels and Limits

### `ax.set_title()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025', loc='center')
plt.show()
````

### `ax.set_xlabel()` and `ax.set_ylabel()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025')
ax.set_xlabel('Month')
ax.set_ylabel('Sales ($)')
plt.show()
````

### `ax.set_ylim()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025')
ax.set_ylabel('Sales ($)')
ax.set_ylim(0, 80000)
plt.show()
````

### `ax.set_xlim()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025: January to June')
ax.set_ylabel('Sales ($)')
ax.set_xlim(-0.5, 5.5)
plt.show()
````

### `ax.set()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set(title='Monthly sales, 2025', ylabel='Sales ($)',
       ylim=(0, 80000))
plt.show()
````

## 3.4 Legends

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2024'])
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales')
ax.set_ylabel('Sales ($)')
plt.show()
````

### `ax.legend()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2024'], label='2024')
ax.plot(sales['month'], sales['sales_2025'], label='2025')
ax.set_title('Monthly sales')
ax.set_ylabel('Sales ($)')
ax.legend()
plt.show()
````

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2024'], label='2024')
ax.plot(sales['month'], sales['sales_2025'], label='2025')
ax.set_title('Monthly sales')
ax.set_ylabel('Sales ($)')
ax.legend(loc='lower right')
plt.show()
````

### Moving the legend outside the chart

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2024'], label='2024')
ax.plot(sales['month'], sales['sales_2025'], label='2025')
ax.set_title('Monthly sales')
ax.set_ylabel('Sales ($)')
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.show()
````

## 3.5 Saving and Showing Charts

### `fig.savefig()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2024'], label='2024')
ax.plot(sales['month'], sales['sales_2025'], label='2025')
ax.set_title('2025 beat 2024 in every month')
ax.set_ylabel('Sales ($)')
ax.set_ylim(bottom=0)
ax.legend(loc='lower right')
fig.savefig('monthly_sales.png', dpi=200, bbox_inches='tight')
plt.show()
````

### `plt.close()`

````python
fig, ax = plt.subplots()
ax.plot(sales['month'], sales['sales_2025'])
ax.set_title('Monthly sales, 2025')
fig.savefig('sales_2025.svg', bbox_inches='tight')
plt.close(fig)
````

## 3.6 Quick Charts with `df.plot()`

### `df.plot()`

````python
sales.plot(x='month', y=['sales_2024', 'sales_2025'])
plt.show()
````

````python
ax = sales.plot(x='month', y=['sales_2024', 'sales_2025'])
ax.set_title('Monthly sales')
ax.set_xlabel('')
ax.set_ylabel('Sales ($)')
ax.legend(['2024', '2025'])
plt.show()
````

````python
fig, ax = plt.subplots(figsize=(5.5, 2.2))
sales.plot(x='month', y='sales_2025', ax=ax, legend=False,
           title='Monthly sales, 2025', ylabel='Sales ($)')
plt.show()
````

### Other kinds of chart

````python
ax = sales.plot.bar(x='month', y='sales_2025', legend=False)
ax.set_title('Monthly sales, 2025')
ax.set_xlabel('')
ax.set_ylabel('Sales ($)')
plt.show()
````
